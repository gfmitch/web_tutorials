## What makes a DataFrame special

OK, so what's so special about a DataFrame? The Pandas dataframe is a powerhouse for managing and manipulating data. It's the Python standard and is expected input and output for many data science and machine learning libraries for Python. So, we *know* we need to learn to use this well. DataFrame *indexes* are one of those things that makes DataFrames special.

### Indexes

Pandas DataFrames actually have two sets of indexes:

- Row indexes - each row can be identified with a non-unique (but we *like* them to be unique) value
- Column indexes (refered to as column names) - each column is identified with a name.

What's so special about indexes? They allow us to access the specific contents of a DataFrame by index value. For example, we can easily *grab* the *name* column from the DataFrame:

```python
statedat['name'].head()
```

Similarly, we can grab a row or rows using indexing (similar to Numpy indexing), including slices.

```python
statedat[0:3]
```

And, as you'd expect, we can be very specific with our subsetting. For example, we can obtain name, region, and income for rows 10-12 if that's all we're interested in:

```python
statedat.loc[10:13,['name', 'region', 'income']]
```

To make the selection above, I used the DataFrames .loc method. When we just use square bracket indexing on the DataFrame instance, e.g. statedat['name'], Pandas expects the argument to be a column name or a list of column names *or* a row index or slice. When we want to select on both the row and column, we must use .loc or .iloc.

### .loc and .iloc

Pandas gives us the ability to access rows and columns using either the index values and column names *as well as* numeric subscripts. Our examples so far have used only the indexes and column names. By default, Pandas index values will be sequential numbers starting at 0. That's what we see in the statedat dataframe. Below, we can see the first few rows of the DataFrame, showing the index values at the left (unlabeled column at the left). We also show the complete index, obtained using the index attribute of the DataFrame.

```python
statedat.head()
```

```python
print([i for i in statedat.index])
```

But we can make the index (and column names) anything we want. Let's change the statedat index to use the state abbreviations:

```python
statedat.index = statedat['abb']
statedat.head()
```

Here we see that the leftmost column now contains the state abbreviations instead of the sequential numbers. Cool, right? It looks a little odd, because now our index and *abb* column contain the same information (we could always drop the abb column). But there's a real advantage to this. Previously, if we wanted to access the data for Oregon, we'd have to use the following:

```python
statedat[statedat['abb'] == 'OR']
```

Now, we can use the 2-character state abbreviation as the subscript:

```python
statedat.loc['OR']
```

&nbsp;

We can even use a slice on the new index. **Note, however, that for *indexes*, unlike Python slices, Pandas *includes* the upper bound**:

```python
statedat[:'CA']
```

We can also use column slices and fancy indexes:

```python
statedat.loc['AZ':'CA']
```

```python
statedat[['abb', 'name', 'center_x', 'center_y']]
```

```python
statedat.iloc[11:21, 3:6]
```

In this case, we used the .iloc property to access rows and columns by position number. **Note that here, when using position number indexing, Pandas follows Python's convention that the upper bound is *excluded* from the result.**

## Pandas Series

If we think of a DataFrame as consisting of a collection of columns of equal length, then we have a good picture of how a DataFrame is implemented. Specifically, a DataFrame consists of a collection of Pandas Series objects, where a Series object is very similar to a Numpy Ndarray. Series behave like numpy arrays in many (most?) respects. However, like DataFrames, Series objects have indexes (but not column names since they are one-dimensional).

We can observe this by extracting a column from our DataFrame and observing its type:

```python
type(statedat['abb'])
```

OK, that wasn't very exciting, but it proved the point. Pandas DataFrame columns are Series. No, what else can we learn about a Series object?

```python
statedat['name'].head()
```

If we look at the first few values in the Series, we'll see that the *extracted* Series object retains the indexes from the DataFrame. Here we see that we can access Series elements by index and by position:

```python
statedat['name']['OR']
```

```python
statedat['name'][10]
```

As with DataFrames, positional slices *exclude* the upper bound

```python
statedat['name'][0:5]
```

but index slices *include* the upper bound

```python
statedat['name'][:'CA']
```

Series objects can be created similarly to Numpy arrays (e.g. using a list)

```python
new_obj = pd.Series(['apple', 'pear', 'watermelon', 'strawberries'])
new_obj
```

Notice that the index here consists of sequential integers. This represents the default because no index values were supplied. Alternatively, we can supply an index when we create the Series:

```python
new_obj = pd.Series(['apple', 'pear', 'watermelon', 'strawberries'],
                    index=['a', 'p', 'w', 's'])
new_obj
```

A nice neat way to construct a Series is using a Python dictionary; the dictionary keys will be used as index values and the dictionary values will be the Series values:

```python
new_obj = pd.Series({'a': 'apple', 'p':'pear', 'w':'watermelon', 's':'strawberries'})
new_obj
```

As with DataFrames, Series objects are very flexible and provide a large number of methods for manipulating the Series.

## Constructing DataFrames

So, how do we create a DataFrame? We already saw one easy method, creating a DataFrame from a .csv file. Alternatively, we can create DataFrames using:

- Lists of lists (each inner list represents a row in the DataFrame)
- Dictionary of lists, specifying column names and/or index values
- Dictionary of Series (each series represents a column in the DataFrame).

### List of lists

```python
df1 = pd.DataFrame([['AK', 'Alaska'],
                    ['AL', 'Alabama'],
                    ['AR', 'Arkansas'],
                    ['AZ', 'Arizona'],
                    ['CA', 'California']
                   ])
df1
```

Note in the result above that the row and column indexes are, by default,  sequential integers.

### Lists of lists specifying column and/or row labels

```python
df1 = df1 = pd.DataFrame([['AK', 'Alaska'],
                    ['AL', 'Alabama'],
                    ['AR', 'Arkansas'],
                    ['AZ', 'Arizona'],
                    ['CA', 'California']
                   ], columns=['Abb', 'Name'])
df1
```

```python
df1 = df1 = pd.DataFrame([['AK', 'Alaska'],
                    ['AL', 'Alabama'],
                    ['AR', 'Arkansas'],
                    ['AZ', 'Arizona'],
                    ['CA', 'California']
                   ], index=['K', 'L', 'R', 'Z', 'A'])
df1
```

```python
df1 = df1 = pd.DataFrame([['AK', 'Alaska'],
                    ['AL', 'Alabama'],
                    ['AR', 'Arkansas'],
                    ['AZ', 'Arizona'],
                    ['CA', 'California']
                   ], columns=['Abb', 'Name'],
                      index=['K', 'L', 'R', 'Z', 'A'])
df1
```

### Dictionary of Series

Here, the dictionary keys become the column labels. The row labels are drawn from the Series indexes.

```python
abbs = pd.Series(['AK', 'AL', 'AR', 'AZ', 'CA'])
names = pd.Series(['Alaska', 'Alabama', 'Arkansas', 'Arizona', 'California'])

pd.DataFrame({'Abb': abbs, 'Name': names})
```

When constructing a DataFrame from Series objects, Pandas *matches* the Series elements using the indexes. In the example below, we construct the two Series objects with *different* indexes. The abbs Series is indexed 0 through 4, whereas the names indexes are 6-10.

```python
abbs = pd.Series(['AK', 'AL', 'AR', 'AZ', 'CA'], index=range(5))
names = pd.Series(['Alaska', 'Alabama', 'Arkansas', 'Arizona', 'California'], index=range(6,11))

pd.DataFrame({'Abb': abbs, 'Name': names})
```

In this result, note the following:
- There is no element 5. Even though the indexes are numbers, they are not necessarily sequential
- There are 10 rows in the DataFrame (rather than 5 previously). The first 5 rows were constructed with the abbs data (they had the lowest index values) while the last 5 were derived from the names
- There are *missing values* for Name in the first 5 rows and for Abb in the last 5 rows.

When constructing this DataFrame, Pandas matched up the two Series objects on the Indexes, with the data associated with each index *merged* to create a row in the resulting DataFrame. Because the first 5 rows had no matching index in the names Series, the Name column values are all set to NaN (and similarly with the Abb values in the last 5 columns).

## Missing values as NaNs

Pandas' treatment of missing values is one of its idiosyncrasies. Most tools that explicitly handle missing values identify them separately from infinity and not a number (Nan). The Pandas designers made an implementation decision to use NaN to represent both missing values and traditional NaN values. This allowed them to streamline their handling of numeric columns, but it *is* something to keep in miind when designing your applications.

### Eliminating missing values

The .dropna DataFrame method can be used to drop missing values from DataFrames. It provides very flexible usage, so you should definitely read the documentation to get a sense of its range of capabilities. The simplest usage (and the frequently desired behavior), is simply to drop all rows where a missing value exists for any column (the rationale being that the observation is incomplete):

```python
dfna = df1
dfna.loc['R','Name'] = pd.NA
dfna
```

```python
dfna.dropna()
```

.dropna() returns an altered DataFrame. But be careful here, by default, .dropna() **does not alter the original DataFrame**.

```python
dfna
```

The output above shows that the original DataFrame, dfna, was *not* altered by the .dropna() call. This is not a mistake. This allows you to make a copy of a DataFrame that excludes the missing values contained in the original. Just assign the .dropna() result to a new variable. There *are* situations, however, when creating this copy can be quite time-consuming (and memory intensive). In these cases, there's an *inplace* argument in the .dropna() call that you can use to instruct Pandas to alter the original DataFrame rather than producing a new one (see the Pandas .dropna() documentation).

There are a number of other useful methods relating to missing values in DataFrames (see the Pandas doc for descriptions):

- isna
- notna
- fillna

## Adding to or modifying an Existing DataFrame

DataFrames are mutable. We can modify data in a DataFrame, just as you would expect:

```python
dfna.loc['R', 'Name'] = 'Arkansas'
dfna
```

We can also delete rows from a DataFrame. Here we delete a row using index values (or column names). When we invoke drop, it *assumes* the index values provided in the call correspond to row index values (unless we tell it otherwise).

```python
dfna.drop('R')
```

Note that, as with .dropna(), .drop() returns a new DataFrame with the changes, rather than updating the current object (unless the inplace=True argument is specified). In the example below, we specify *axis=1* to tell Pandas to drop a column to drop (axis=0, the default, specifies row deletion). This works fine, but I don't really like this version. It's too cryptic.

```python
dfna.drop('Abb', axis=1)
```

I prefer the following alternative for specifying column deletion.

```python
dfna.drop(columns='Abb')
```

We can also delete rows that satisfy a condition (*all* rows satisfying the condition):

```python
dfna.drop(dfna[dfna['Abb'] == 'AZ'].index)
```

The code above bears explanation. Executing from the inside out will illustrate. The innermost code extracts the rows that satisfy the condition. .drop() won't know how to operate on this, because it expects a set of index values to tell it which rows to drop.

```python
dfna[dfna['Abb'] == 'AZ']
```

So, we add .index to the DataFrame subset; .index will return the index values for the subset.

```python
dfna[dfna['Abb'] == 'AZ'].index
```

Here's another example showing the indexes matching the condition that the abbreviation $\le$ 'A'

```python
dfna[dfna['Abb'] <= 'Aa'].index
```

So why did $\le$ 'Aa' return the indexes for the first four rows? Because, to a computer 'Z' $<$ 'a'. In other words, upper case letters have a lower sort value than lower case letters. Ultimately, we see that .drop() is *always* using index values for dropping the rows, but we have a variety of ways that we can obtain those indexes.

```python
dfna.drop(dfna[dfna['Abb'] == 'AZ'].index)
```

## Adding Columns and Rows to DataFrames
Adding data to DataFrames is *relatively* easy (there are occasionally tricky exceptions). To illustrate, let's start by adding Oregon to our DataFrame. We'll use the .loc proprty to let Pandas know that we're adding a row via an index. Note that an easy way to supply the values is as a dictionary; using a dictionary for the new data allows the new data to appear in any order

```python
df2 = df1.copy()
df2.loc['O'] = {'Name': 'Oregon', 'Abb': 'OR'}
df2
```

Alternatively, we could supply the new data, in column order:

```python
df2.loc['W'] = ['WA', 'Washington']
df2
```

We could also add multiple rows at once, but we'll have to use .concat; .loc and .iloc can be used to add a single row, but not multiple rows at once. The benefit of using .loc (or .iloc) is that the current DataFrame is updated.

```python
df2 = df1.copy()

# construct DataFrane with new rows and set indexes
new_df = pd.DataFrame({'Abb': pd.Series(['ID', 'MT']), 'Name': pd.Series(['Idaho', 'Montana'])})
new_df.index = ['I', 'M']

# Now, concatenate the DataFrames
combined = pd.concat([df2, new_df])
combined
```

The Pandas .concat() function (it's a function, not an instance method) returns a new DataFrame. Using .concat() is easy (line 8 above) when you are trying to append one DataFrame to another. The first argument to .concat() is a list of the Pandas objects you wish to concatenate. This allows you to concatenate as many different Pandas objects as you like.

If your new data is not already in a DataFrame, it's probably easiest to construct a new DataFrame with the new data rather than using other approaches. The .concat() method is very flexible and can be used to add both rows and columns to DataFrame. You should review the .concat() documentation for more details, there's a lot there.

## .join() and .merge()

The .join() and .merge() methods (of DataFrame) allow you to combine DataFrames in a method analogous to SQL joins and merges. For example, we can add a new column indicating whether or note a state has a coastline:

```python
coastal = pd.Series([True, True, False, False, True], index=['AK', 'AL', 'AR', 'AZ', 'CA'], name='Coastal')
df1.join(coastal, on='Abb')
```

This is really nice! Pandas will match up the rows using the *index* values of the argument and either the index values of the original DataFrame or, as in this example, the values in a column (specified using the *on* argument). Note that the *name* property of the Series is used as the column name in the joined DataFrame.

Similarly, we can use merge. However, *merge* gives us even more flexibility in how Pandas will match the rows in the inputs; we can use either columns or indexes for the merge keys in both the *left* and *right* DataFrames (the left DataFrame is the object used to invoke the .merge method).

```python
coastal = pd.Series([True, True, False, False, True], index=['AK', 'AL', 'AR', 'AZ', 'CA'], name='Coastal')
df1.merge(coastal, left_on='Abb', right_index=True)
```
