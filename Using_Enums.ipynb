# Using Enumerations (enums)

An enum represents a finite set of related values. Enums are useful
for ensuring that a variable assumes only valid values. For example,
let's say we had a variable to hold the current day of the week. The only 
valid values are:

-   Sunday
-   Monday
-   Tuesday
-   Wednesday
-   Thursday
-   Friday
-   Saturday

If we use a standard string variable, we could assign a value as follows:

```python
day = "Tuesday"
```

Of course, because *day* is just a string variable, we could also assign
an incorrect (e.g. from a typo) or completely invalid value:

```python
day = "Bluesday"
```

The fundamental problem with this is that we now have an invalid day of the week. Any 
logic that relies on *switching* on day (i.e. if-elifs that implement different logic
by day of the week) will be broken by our invalid day, *Bluesday*.

## Enums to the rescue
The solution to this problem is to use an enumeration (enum). Using the enums, we specify
the finite set of values for the enumeration. If we then always use the enum in our code,
we eliminate the possibility of an invalid value.

## Example

To define an enum, we first import the Enum class and then declare a new class that
inherits Enum.

```python
from enum import Enum

class DOW(Enum):
    SUNDAY = 0
    MONDAY = 1
    TUESDAY = 2
    WEDNESDAY = 3
    THURSDAY = 4
    FRIDAY = 5
    SATURDAY = 6
```

The enum declaration is very straightforward; we define the set of possible *names* and
their corresponding *values*. By convention, the enum *names* are always specified in all
capital letters.

To use the enum, we assign one of the *names* to a variable which will result in the 
corresponding value being assigned to the variable.

```python
day = DOW.TUESDAY
```

The enum allows us to make assignments and do comparisons using
using clearly defined values, but with much greater efficiency
than assigning and comparing string literals

Comparisions are much clearer than day == 2 and more efficient than
day == 'Tuesday'

```python
print(day == DOW.TUESDAY)
print(day is DOW.TUESDAY)
```

When we print the variable, we see its name, TUESDAY, rather than the
actual value of 2 held by the variable

```python
print(day)
```

A string representation of the enum *name* is accessible using the 
enum variable's *.name* property.

```python
print(day.name)
```

We can access the underlying value that is actually stored in the 
variable, which can be useful for efficiently relating names and
operational values. For example, if we wanted to find out the 
day # (i.e. the number of the day of the week), for the variable 
*day*, we can simply access its *value*.

```python
print(day.value)
```

Using an enum, we can also find the *name* that corresponds
to a value. For example, we can use the enum to find the name
of the nth day of the week.

```python
print(DOW(0))
print(DOW(2))
```

There's a lot more that can be done with enums. For example, since
an enum is just a special Python class, we could add attributes
and methods to the enum.

```python
class DOW(Enum):
    SUNDAY = 0
    MONDAY = 1
    TUESDAY = 2
    WEDNESDAY = 3
    THURSDAY = 4
    FRIDAY = 5
    SATURDAY = 6
    
    def next(self):
        next_day = self.value + 1
        
        if next_day > DOW.SATURDAY.value:
            return DOW(0)
        else:
            return DOW(next_day)
    
day = DOW.TUESDAY
print(day)
print(day.next())
```

## One Caveat
Enums are worth getting to know and experimenting with. Unfortunately, because
Python is an implicitly typed language (i.e. we do not explicitly declare 
variable types), there is no way to enforce the use of the enum.

For example, let's say we start out in one area of our code using the enum.

```python
day = DOW.FRIDAY
print(type(day))
```

However, later we forget that we were using an enum, and use the following
assignment:

```python
day = 'Sunday'
print(type(day))
```

As you can see, Python does not protect us from ourselves. When we did
the second assignment, Python replaced our original enum object with
a string object and we succeeded in breaking all subsequent code that
expects day to be of type DOW.

This is one of the downsides of implicit typing. A language like Java,
which requires explicit type declarations, would not allow us to make
the second assignment; type mismatch errors are caught at compile time
(which is a *good* thing).
