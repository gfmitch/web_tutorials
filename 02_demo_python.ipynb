```python
# load the necessary Pandas and Matplotlib libraries
import pandas as pd
import matplotlib.pyplot as plt

# load data from CSV file
baseball = pd.read_csv("baseball.csv")

# select subset from full dataset
sub = baseball[baseball.id == 'bondsba01']

# create simple X,Y scatterplot
plt.scatter(sub.year, sub.hr)

```

```python
import numpy as np

pct_err_dmd = np.arange(-0.5, 2.01, 0.05)

pct_cost_prem = 0.5 * (np.sqrt(1.0 + pct_err_dmd) + np.sqrt(1.0 / (1.0 + pct_err_dmd))) - 1
# plt.scatter(pct_err_dmd,
#             pct_cost_prem,
#             marker=None,
#             linewidths=1)
plt.plot(pct_err_dmd, pct_cost_prem)
plt.xlabel('% Demand Estimate Error')
plt.ylabel('% Cost Premium')
plt.title('EOQ Sensitivity to Demand Estimate Error')
```
