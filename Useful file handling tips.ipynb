Python provides **many** useful utilities for interacting with the operating system. Some of these
involve obtaining directory listings and determining whether a directory entry represents a file
or a subdirectory (i.e. folder).


## Obtaining a list of all *entries* in a directory

There are a number of ways to obtain a directory listing in Python, each utilizes an existing
package making the task easy.

### os.listdir

One of the simplest ways of obtaining a directory listing is to use the listdir(path) method
found in the os package. This function will return the entries in the directory identified
by path:
- If no path is specified, defaults to the current directory
- If a path is specified, lists the directory specified by path
    - path may be a relative or absolute path
    - '.' and '..' refer to the current and parent directories respectively, according to the UNIX convention

Here's an example that lists all of the entries in the current directory:

```python
import os
os.listdir()
```

The following lists all of the entries in the .idea subdirectory of the current directory:

```python
os.listdir('.idea')
```

And the following lists the entries in the *parent* directory of the current directory:

```python
os.listdir('..')
```

## Looking strictly for files

As we noted above, listdir() returned *all* entries in a directory, including subdirectories.
If we want to find only the *files* to avoid attempting to open a subdirectory as a file,
we can use os.path.isfile(). Here's an example that determines whether each entry
is a file:

```python
import pandas as pd

entries = os.listdir()
files = [{'Entry': e,
         'Type': 'File' if os.path.isfile(e) else 'Other'} for e in entries]

print(pd.DataFrame(files))
```

In the example above, we listed the Type as 'Other' if the entry is not a file because
files and directories are not the only types of entries (e.g. symbolic links, etc.).

Similarly, we could use the isdir function:

```python
import pandas as pd

entries = os.listdir()
files = [{'Entry': e,
         'Type': 'Dir' if os.path.isdir(e) else 'Other'} for e in entries]

print(pd.DataFrame(files))
```

## Selecting *only* the files in a directory

If we're looking to find only the files in a directory, we can add a filter to the our list comprehension. The
following obtains the complete list of directory entries in the current directory, but then keeps only those
that test as *files*:

```python
entries = os.listdir()
files = [e for e in entries if os.path.isfile(e)]

print(files)
```

## Matching filename patterns

Directories often contain more files than just those we are looking for. For example, we may want to find all
of the .txt files, but the directory additionally Word documents, Python program files, etc.

We have a couple of options here. Given a list of files like we produced in the above example, we can use the
fnmatch function to filter out only those that match on a particular filename pattern:

```python
import fnmatch

entries = os.listdir()
files = [e for e in entries if os.path.isfile(e)]
csvfiles = [f for f in files if fnmatch.fnmatch(f, '*.csv')]
print(csvfiles)
```

We can even incorporate both selections into a single statement:

```python
import fnmatch

entries = os.listdir()
csvfiles = [e for e in entries if os.path.isfile(e) and fnmatch.fnmatch(e, '*.csv')]
print(csvfiles)
```

We can also select on more than just the file extension, for example, to select all of the boiler sample files,
we could use the following:

```python
import fnmatch

entries = os.listdir()
csvfiles = [e for e in entries if os.path.isfile(e) and fnmatch.fnmatch(e, 'boiler_sample_*.csv')]
print(csvfiles)
```

## glob

The glob module offers an alternative to using listdir and fnmatch. The following example illustrates using
glob to acquire a list of files matching the pattern 'boiler_sample_*.csv'.

```python
import glob

pathnames = glob.glob( './boiler_sample_*.csv', recursive=False)

for p in pathnames:
    print(p)
```

A significant difference between listdir and glob is that glob returns *pathnames* (i.e. filenames containing
the path). The basename function can be used to extract just the filename if that's all we need:

```python
files = [os.path.basename(f) for f in pathnames]
print(files)
```

And if we want to ensure that we are extracting names only for *files* (rather than directories or symbolic
links), then we can add a filter to the list comprehension:

```python
files = [os.path.basename(f) for f in pathnames if os.path.isfile(f)]
for f in files:
    print(f)
```
